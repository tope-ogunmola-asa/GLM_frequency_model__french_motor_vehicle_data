# 04 - Overdispersion and Train/Test Split

```python
# Import LIbraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import fetch_openml
from sklearn.model_selection import train_test_split
import statsmodels.api as sm
import statsmodels.formula.api as smf
from scipy.stats import chi2

# Display all columns
pd.set_option('display.max_columns', None)
```

```python
# Read merged data from Parquet file
fmv_df = pd.read_parquet("fmv_df.parquet")

fmv_df.head()
```

```python
# Examine the frequency distribution for overdispersion
mean_claims = fmv_df["ClaimNb"].mean()
var_claims = fmv_df["ClaimNb"].var()

print("Mean:", mean_claims)
print("Variance:", var_claims)

# tolerance for "approximately equal"
tolerance = 0.1 * mean_claims  

if var_claims > mean_claims:
    print("Distribution of claims is overdispersed since the variance is greater than the mean.")
elif abs(mean_claims - var_claims) <= tolerance:
    print("Distribution of claims possibly follow Poisson distribution.")
else:
    print("Distribution of claims is underdispersed.")

############################## Takeaway ##############################
# The distribution of claims suggests a Negative Binomial distribution may be appropriate
# since the Negative Binomial distribution adds a flexible dispersion parameter.
```
